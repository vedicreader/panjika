# skill

> the skill, for a harness that reads skills from packages

`SKILL.md` is in the package. `install` writes it into the skill folder of each agent. This module publishes the same text through `[project.entry-points.pyskills]`. It reads the body from the packaged file, so the text is not written two times.

```
#| export
from panjika.harness import skill_body

__doc__ = skill_body()
```

```
from fastcore.test import test_eq
from panjika.harness import frontmatter, skill_md

import panjika.skill
meta, body = frontmatter(skill_md())
test_eq(panjika.skill.__doc__, skill_body())
assert panjika.skill.__doc__.startswith(meta['description'].split('.')[0])
assert '---' not in panjika.skill.__doc__.split('\n')[0]
assert body in panjika.skill.__doc__
```
