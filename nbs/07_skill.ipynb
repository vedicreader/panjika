# skill

> the skill, published to a harness that reads them from packages

`SKILL.md` lives in the package. `panjika install` writes it into each agent's skill folder, and this module publishes the same text through `[project.entry-points.pyskills]` for a harness that loads skills from installed packages instead of from disk.

A pyskill body is the module docstring, so it is assigned from the packaged file rather than written twice.

```
#| export
from panjika.harness import skill_body

__doc__ = skill_body()
```

```
from fastcore.test import test_eq

import panjika.skill
test_eq(panjika.skill.__doc__, skill_body())
assert panjika.skill.__doc__.lstrip().startswith('# panjika')
assert '---' not in panjika.skill.__doc__.split('\n')[0]
```
